```python
import pandas as pd

df = pd.read_csv("cleveland.csv")

df["fbs"] = df["fbs"].astype(bool)
df["exang"] = df["exang"].astype(bool)
df["num"] = df["num"] != 0

for col, dtype in zip(df.columns, df.dtypes):
    if dtype == "float64":
        df[f"{col}_s"] = (df[col] - df[col].mean()) / df[col].std()

df
```
